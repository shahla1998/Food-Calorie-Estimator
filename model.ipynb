```python
import pandas as pd
import numpy as np
import cv2
from tensorflow.keras.models import load_model

df = pd.read_csv('/Users/mishabulhaq/Desktop/Food_Calories_Estimator/Nutrition Sheet - Sheet1 (1).csv') 

model = load_model('food_model.h5')

class_names = ['Egg', 'Meat', 'Rice', 'Seafood', 'Vegetable-Fruit']

def predict_food_from_image(image_path):
    img = cv2.imread(image_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (224, 224))
    img = img / 255.0
    img = np.expand_dims(img, axis=0)
    
    predictions = model.predict(img)
    predicted_index = np.argmax(predictions)
   
    predicted_food_name = class_names[predicted_index]
    return predicted_food_name

def estimate_food_calories(food_name): 
    item = df[df['Food Items'].str.lower() == food_name.lower()] 
    
    if not item.empty: 
        protein = item['Protein'].values[0]
        carbs = item['Carbs'].values[0] 
        fat = item['Fat'].values[0] 
        
        pred_cal = (protein * 4) + (carbs * 4) + (fat * 9) 
        
        print(f"--- {food_name.upper()} ---") 
        print(f"Protein: {protein}g | Carbs: {carbs}g | Fat: {fat}g") 
        print(f"Estimated Calories: {pred_cal:.2f} kcal") 
    else:
        print("Food is not available in the list!")


image_path = '/Users/mishabulhaq/Desktop/Food_Calories_Estimator/dataset/train/Seafood/156.jpg' 
predicted_name = predict_food_from_image(image_path)

estimate_food_calories(predicted_name)
```
